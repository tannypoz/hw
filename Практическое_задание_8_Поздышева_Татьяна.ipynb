# **Работа с API**

Шаг 1: Получение информации обо всех книгах

```python
import requests
import pandas as pd
```

```python
# Базовый URL API
api_base_url = 'https://www.anapioficeandfire.com/api/'
```

```python
book_response = requests.get(api_base_url + 'books').json()

print("Количество книг:", len(book_response))

print("Первые три книги:\n", book_response[:3])  # показываем первые три книги

df_books = pd.DataFrame(book_response)
print("\nDataFrame с книгами:\n")
df_books.head()
```

Шаг 2: Получение информации обо всех домах Вестероса

```python
house_response = requests.get(api_base_url + 'houses').json()
print("\nПервые три дома:\n", house_response[:3])  # показываем первые три дома

df_houses = pd.DataFrame(house_response)
print("\nDataFrame со всеми домами:\n")
df_houses.head()
```

Шаг 3: Получение информации о домах с девизом

```python
homes_with_titles_response = requests.get(api_base_url + 'houses/?hasTitles=true').json()
print("\nПервые три дома с девизами:\n", homes_with_titles_response[:3])  # показываем первые три дома с девизами

df_houses_with_titles = pd.DataFrame(homes_with_titles_response)
print("\nDataFrame с домами, имеющими девиз:\n")
df_houses_with_titles.head()
```

Шаг 2: Извлечение первых 10 записей из таблицы rnc_database

```python
# Получение 10 строк из таблицы rnc_database
with conn.cursor() as cursor:
    cursor.execute('SELECT * FROM rnc_database LIMIT 10')  # Используем LIMIT 10 для ограничения числа строк
    result = cursor.fetchall()
```

```python
# Преобразование в Pandas DataFrame
df_full = pd.DataFrame(result, columns=[col.name for col in cursor.description])
print("Полный DataFrame с первыми 10-ю строками:")
print(df_full)
```

Шаг 3: Сохраняем выбранные колонки

```python
# Выборка необходимых столбцов
with conn.cursor() as cursor:
    cursor.execute('''
        SELECT display_name, num_sequences, num_organisms, url
        FROM rnc_database
        LIMIT 10
    ''')
    selected_result = cursor.fetchall()
```

```python
# Преобразование выбранного подмножества в новый DataFrame
df_selected = pd.DataFrame(selected_result, columns=["display_name", "num_sequences", "num_organisms", "url"])
print("\nDataFrame с необходимыми столбцами (display_name, num_sequences, num_organisms, url):")
print(df_selected)
```
