Шаг 4: Реализация интерфейса для предсказания с помощью Ipywidgets

```python
import ipywidgets as widgets
from IPython.display import display
```

```python
# Форма для ввода данных
square_widget = widgets.FloatText(value=50.0, description='Общая площадь:')
room_widget = widgets.IntSlider(min=1, max=5, value=2, description='Комнат:')
floor_widget = widgets.IntSlider(min=1, max=10, value=1, description='Этаж:')  # Новая форма для ввода этажа
predict_button = widgets.Button(description='Прогнозировать')
output = widgets.Output()
```

```python
# Логика предсказания
def predict_price(b):
    square = float(square_widget.value)
    room_count = int(room_widget.value)
    floor = int(floor_widget.value)  # Берём значение этажа
    pred_input = [[square, room_count, floor]]  # Три признака: площадь, комнаты, этаж
    prediction = model.predict(pred_input)[0]
    with output:
        print(f"Прогнозируемая стоимость: {round(prediction)} рублей")

predict_button.on_click(predict_price)
```

```python
# Собираем форму вместе
ui = widgets.VBox([square_widget, room_widget, floor_widget, predict_button, output])
display(ui)
```
